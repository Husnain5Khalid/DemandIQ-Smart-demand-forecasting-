## Section: 1 : Build  a baseline Model

```python
import pandas as pd
from pathlib import Path

processed_path = Path("../data/processed")

train_data = pd.read_csv(
    processed_path / "train_data.csv"
)

valid_data = pd.read_csv(
    processed_path / "valid_data.csv"
)
```

```python
train_data["date"] = pd.to_datetime(train_data["date"])
valid_data["date"] = pd.to_datetime(valid_data["date"])
```

```python
TARGET = "sales"

X_train = train_data.drop(columns=[TARGET])
y_train = train_data[TARGET]

X_valid = valid_data.drop(columns=[TARGET])
y_valid = valid_data[TARGET]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
X_train = X_train.drop(columns=["date"])
X_valid = X_valid.drop(columns=["date"])
```

```python
categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

numeric_features = X_train.select_dtypes(
    exclude=["object"]
).columns.tolist()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_valid_processed = preprocessor.transform(X_valid)
```
